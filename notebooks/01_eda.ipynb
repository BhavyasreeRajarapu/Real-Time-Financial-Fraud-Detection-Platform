```python
import pandas as pd
df=pd.read_csv("../dataset/paysim.csv")
print(df.shape)
df.head()
```

```python
print(df["isFraud"].value_counts())
print(df["isFraud"].mean())
```

```python

fraud_by_type = df.groupby("type")["isFraud"].agg(["count", "sum"])
fraud_by_type.columns = ["total_transactions", "fraud_transactions"]
print(fraud_by_type)
```

```python
print(df["step"].min(), df["step"].max())
print(df.groupby("step")["isFraud"].sum().describe())
```

```python
print(df.isna().sum())
print(df["nameOrig"].nunique(), df["nameDest"].nunique())
```

```python
fraud = df[df["isFraud"] == 1]
print(fraud[["amount", "oldbalanceOrg", "newbalanceOrig"]].describe())
print(df[df["isFraud"] == 0][["amount", "oldbalanceOrg", "newbalanceOrig"]].describe())
```
